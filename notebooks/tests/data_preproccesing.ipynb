```python
# Cell C1: VolumeDataset
# Dataset class that loads 3D volumes and applies transforms.
# Integrates with data_utils/load_data.py for volume loading.

def remap_image(image: torch.Tensor, new_min: float = 0.0, new_max: float = 100.0) -> torch.Tensor:
    """Remap tensor values to [new_min, new_max] range.
    
    This is the per-batch normalization used in CellSeg3D training.
    Ref: worker_training.py lines 593-597 (uses utils.remap_image).
    """
    old_min = image.min()
    old_max = image.max()
    if old_max - old_min < 1e-8:
        return torch.full_like(image, new_min)
    return (image - old_min) / (old_max - old_min) * (new_max - new_min) + new_min

```
